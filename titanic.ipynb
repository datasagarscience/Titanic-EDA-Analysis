```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the dataset
df = pd.read_csv(r"C:\Users\hp\Desktop\titanic dataset\tested.csv")
print("--- First 5 Rows of the Dataset ---")
display(df.head())
```

```python
# Check data info and missing values
print("\n--- Dataset Information ---")
print(df.info())
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# Data Cleaning and Handling Missing Values.

# Fill missing values in 'Age' with the median age using NumPy/Pandas.
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)

# Fill missing values in 'Fare' with the mean fare
mean_fare = df['Fare'].mean()
df['Fare'] = df['Fare'].fillna(mean_fare)
print("\nMissing values successfully handled!")
```

```python
# Statistical Analysis using Pandas and NumPy

print("\n--- Statistical Insights ---")
print(f"Total Passengers: {len(df)}")
print(f"Total Survived: {df['Survived'].sum()}")
print(f"Average Age: {df['Age'].mean():.2f} years")
print(f"Maximum Fare: {df['Fare'].max():.2f}")
print(f"Minimum Fare: {df['Fare'].min():.2f}")
```

```python
# Group by Passenger Class (Pclass) to see survival counts
print("\n--- Class-wise Survival Count ---")
print(df.groupby('Pclass')['Survived'].sum())
```

```python
# Custom NumPy operation.
df['Log_Fare'] = np.log(df['Fare'] + 1)
```

```python
# Data Visualization: Survival Count by Gender
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Sex', hue='Survived', palette='Set2')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```
